```python
# Flipkart Product Scraper 

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# -------------------------------
# Step 1: Set the product keyword
# -------------------------------
search_query = "laptop"  # Change to "mobile", "shoes", "watch", etc.
base_url = "https://www.flipkart.com/search?q=" + search_query

# -------------------------------
# Step 2: Prepare empty lists
# -------------------------------
product_names = []
prices = []
ratings = []

# -------------------------------
# Step 3: Loop through first 5 pages
# -------------------------------
for page in range(1, 6):  # Change 6 to 11 for 10 pages
    print("Scraping page:", page)
    url = base_url + "&page=" + str(page)
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    # -------------------------------
    # Step 4: Find product info
    # -------------------------------
    name_tags = soup.find_all("div", class_="KzDlHZ")
    price_tags = soup.find_all("div", class_="Nx9bqj _4b5DiR")
    rating_tags = soup.find_all("div", class_="XQDdHH")

    # -------------------------------
    # Step 5: Save data in lists
    # -------------------------------
    for i in range(len(name_tags)):
        product_names.append(name_tags[i].text if i < len(name_tags) else "N/A")
        prices.append(price_tags[i].text if i < len(price_tags) else "N/A")
        ratings.append(rating_tags[i].text if i < len(rating_tags) else "N/A")

    time.sleep(2)  # wait 2 seconds to avoid blocking

# -------------------------------
# Step 6: Create DataFrame
# -------------------------------
data = pd.DataFrame({
    "Product Name": product_names,
    "Price": prices,
    "Rating": ratings
})

# -------------------------------
# Step 7: Save CSV
# -------------------------------
file_name = "flipkart_" + search_query + "_data.csv"
data.to_csv(file_name, index=False)

# -------------------------------
# Step 8: Summary
# -------------------------------
print("Scraping finished!")
print("Total products scraped:", len(data))
print("CSV file saved as:", file_name)
print(data.head(10))
```
